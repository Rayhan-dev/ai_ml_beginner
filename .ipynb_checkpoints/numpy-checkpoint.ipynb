```python
import numpy as np
rng = np.random.default_rng(seed=1) #random generation

ar = rng.integers(low=0,high=100,size=(3,3))
print(ar)

ne = np.where(ar <50 , 0 , ar) # filtering without flat
print(ne)

print(ar[ar % 2 == 0]) # flat filtering
```
