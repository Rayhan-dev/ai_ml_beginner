```python
import numpy as np
rng = np.random.default_rng(seed=1) #random generation

ar = rng.integers(low=0,high=100,size=(3,3))
print(ar)

ne = np.where(ar <50 , 0 , ar) # filtering without flat
print(ne)

print(ar[ar % 2 == 0]) # flat filtering
```

```python
st = { 1, 2, 3 }

arr = np.array(list(st))
print(arr.dtype) 
arr = arr.astype(np.int16)
print(arr) 
print(type(arr)) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim) 
```

```python
# dictionary 

dc = { 'a' : 10 , 'b' : 20 , 'c' :30 }

keys = dc.keys()
values = dc.values()
items = dc.items()
# print(keys)
# print(values)
# print(items)
```
